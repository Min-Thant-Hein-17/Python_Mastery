```python
import numpy as np
```

```python
# Set random seed for reproducibility
np.random.seed()
```

```python
#Dataset (Generate Random Data):

number_of_movies = 500

genres_list = ["Action", "Comedy", "Drama", "Thriller", "Sci-Fi"]

movie_ids = np.arange(1, number_of_movies + 1)
ratings = np.random.randint(1, 6, size=number_of_movies)
views = np.random.randint(100, 10001, size=number_of_movies)
genres = np.random.choice(genres_list, size=number_of_movies)
```

# Task 1: Data Preparation

```python
# Calculate the total rating score for each movie
total_rating_scores = ratings * views

# Classify each movie based on its average rating
low_rated = movie_ids[ratings < 2.5]
medium_rated = movie_ids[(ratings >= 2.5) & (ratings <= 4)]
high_rated = movie_ids[ratings > 4]
```

```python
print(total_rating_scores)
```

```python
print(low_rated)
print(medium_rated)
print(high_rated)
```

# Task 2: Data Analysis

```python
# Find the average view count for each genre
unique_genres = np.unique(genres)
avg_views_by_genre = {genre: np.mean(views[genres == genre]) for genre in unique_genres}

# Calculate the total view count for each rating category
total_views_low = views[ratings < 2.5].sum()
total_views_medium = views[(ratings >= 2.5) & (ratings <= 4)].sum()
total_views_high = views[ratings > 4].sum()

# Find the top 5 most-watched movies
top_5_most_watched = movie_ids[np.argsort(views)[-5:]][::-1]

# Determine the percentage of movies in each rating category
percent_low_rated  = len(low_rated) / number_of_movies * 100
percent_medium_rated = len(medium_rated) / number_of_movies * 100
percent_high_rated = len(high_rated) / number_of_movies * 100
```

# Task 3: Advanced Analysis

```python
# Calculate the standard deviation of ratings in each genre
std_rating_by_genre = {genre: np.std(ratings[genres == genre]) for genre in unique_genres}

# Find the most common genre among the top 10 most-watched movies
top_10_most_watched_genres = genres[np.argsort(views)[-10:]]
most_common_genre_top_10 = np.bincount([np.where(unique_genres == genre)[0][0] for genre in top_10_most_watched_genres]).argmax()
most_common_genre_top_10 = unique_genres[most_common_genre_top_10]
```
