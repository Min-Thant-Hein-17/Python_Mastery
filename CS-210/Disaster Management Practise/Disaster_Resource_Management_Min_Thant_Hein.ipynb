```python
affected_regions = ["Mandalay", "Mon", "Yangon"]
```

```python
resource_inventory = {
    "Food": 1000,
    "Water": 2000,
    "Medicine": 500,
}
```

```python
resource_distribution = {}
```

```python
selected_region = "Mandalay" # Distribute resoures
selected_resource = "Food"
quantity_to_distribute = 200
```

```python
# Check resource availability and distribute
if selected_resource in resource_inventory and resource_inventory[selected_resource] >= quantity_to_distribute:
    resource_inventory[selected_resource] -= quantity_to_distribute
    if selected_region not in resource_distribution:
        resource_distribution[selected_region] = {}
    resource_distribution[selected_region][selected_resource] = resource_distribution[selected_region].get(selected_resource, 0) + quantity_to_distribute
else:
    print(f"Not enough {selected_resource} to distribute to {selected_region}.")
```

```python
region_status = {
    "Mandalay": "Critical",
    "Mon": "Stable",
    "Yangon": "Recovered"
}
```

```python
region_status["Mandalay"] = "Stable" #update region status
region_status["Mon"] = "Recovered"
region_status["Yangon"] = "Recovered"
```

```python
# Generate report
print(" Summary Report ")
print("\nResources Distributed:")
for region, resources in resource_distribution.items():
    print(f"{region}:")
    for resource, qty in resources.items():
        print(f"  {resource}: {qty} ")

print("\nRegion Status:")
for region, status in region_status.items():
    print(f"{region}: {status}")

print("\nRemaining Inventory:")
for resource, qty in resource_inventory.items():
    print(f"{resource}: {qty} ")
```
