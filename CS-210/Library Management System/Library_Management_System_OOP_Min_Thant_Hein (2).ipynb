```python
from abc import ABC, abstractmethod

class Book:
    def __init__(self, title, author, isbn, available_copies):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.available_copies = available_copies

    def __str__(self):
        return f"{self.title} by {self.author} (ISBN: {self.isbn}) - Available copies: {self.available_copies}"

    def borrow_book(self):
        if self.available_copies > 0:
            self.available_copies -= 1
            return True
        else:
            return False

    def return_book(self):
        self.available_copies += 1
```

```python
# Abstract Base Class for Members
class Member(ABC):
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    @abstractmethod
    def borrow_limit(self):
        pass

    def borrow(self, book):
        if len(self.borrowed_books) < self.borrow_limit():
            if book.borrow_book():
                self.borrowed_books.append(book)
                return f"{self.name} successfully borrowed '{book.title}'."
            else:
                return f"No copies of '{book.title}' available for borrowing."
        else:
            return f"{self.name} has reached the borrowing limit of {self.borrow_limit()} books."

    def return_book(self, book):
        if book in self.borrowed_books:
            book.return_book()
            self.borrowed_books.remove(book)
            return f"{self.name} successfully returned '{book.title}'."
        else:
            return f"{self.name} did not borrow '{book.title}'."

# RegularMember inherits from Member
class RegularMember(Member):
    def borrow_limit(self):
        return 2  # Regular members can borrow up to 2 books

# PremiumMember inherits from Member
class PremiumMember(Member):
    def borrow_limit(self):
        return 5  # Premium members can borrow up to 5 books
```

```python
class Librarian(Member):
    def borrow_limit(self):
        return float('inf')  # Librarians can borrow an unlimited number of books

    def add_book(self, library, book):
        library.add_book(book)
        return f"Librarian {self.name} added '{book.title}' to the library."
```

```python
class Library:
    def __init__(self):
        self.books = {}
        self.members = {}

    def add_book(self, book):
        if book.isbn in self.books:
            self.books[book.isbn].available_copies += book.available_copies
        else:
            self.books[book.isbn] = book

    def register_member(self, member):
        self.members[member.name] = member

    def check_out(self, member, book):
        if book.isbn in self.books:
            return member.borrow(self.books[book.isbn])
        else:
            return f"'{book.title}' is not available in the library."

    def return_book(self, member, book):
        if book.isbn in self.books:
            return member.return_book(self.books[book.isbn])
        else:
            return f"'{book.title}' does not belong to this library."
```

```python
# Example Usage
library = Library()
```

```python
# Adding Books
book1 = Book("1984", "George Orwell", "1234567890", 3)
book2 = Book("To Kill a Mockingbird", "Harper Lee", "2345678901", 2)
library.add_book(book1)
library.add_book(book2)

# Registering Members
alice = RegularMember("Alice")
bob = PremiumMember("Bob")
john = PremiumMember("John")
library.register_member(alice)
library.register_member(bob)
library.register_member(john)
```

```python
# Borrowing Books
print(alice.borrow(book1))  # Alice borrows "1984"
print(bob.borrow(book1))    # Bob borrows "1984"
print(bob.borrow(book2))    # Bob borrows "To Kill a Mockingbird"
print(bob.borrow(book2))    # Bob tries to borrow "To Kill a Mockingbird" again (no copies available)
print(john.borrow(book1))   # John borrows "1984"

# Returning Books
print(alice.return_book(book1))  # Alice returns "1984"
print(bob.borrow(book1))         # Now Bob can borrow "1984".
```
