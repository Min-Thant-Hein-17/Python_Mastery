```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
choices = ['rock', 'paper', 'scissors']
```

```python
def get_computer_choice():
  computer_choice = random.choice(choices)
  return computer_choice
```

```python
print(get_computer_choice())
```

```python
def get_player_choice():
  player_choice = input("Enter a choice (rock, paper, scissors): ")
  if player_choice not in choices:
    print("Invalid choice! Please try again.")
    return get_player_choice()
  else:
    return player_choice
```

```python
def determine_winner(computer_choice, player_choice):
  if computer_choice == player_choice:
    return f"Computer chose {get_computer_choice()}, It's a tie!"
  elif computer_choice == 'rock':
    if player_choice == 'scissors':
      return f"Computer chose {get_computer_choice()}, Computer wins!"
    else:
      return f"Computer chose {get_computer_choice()}, Player wins!"
  elif computer_choice == 'paper':
    if player_choice == 'rock':
      return f"Computer chose {get_computer_choice()}, Computer wins!"
    else:
      return f"Computer chose {get_computer_choice()}, Player wins!"
  elif computer_choice == 'scissors':
    if player_choice == 'paper':
      return f"Computer chose {get_computer_choice()}, Computer wins!"
    else:
      return f"Computer chose {get_computer_choice()}, Player wins!"
  else:
    return "Invalid input! You have not entered rock, paper or scissors, try again."
```

# Note: There is one thing that computer makes two predictions, since i have included "f"Computer chose {get_computer_choice()},"". However, it's still a fair game and the prediction is randomn. So, I kept it by not solving! The Purpose of including that text line is to show the prediction/ the choice that computer makes, making sure it's a fair game and we, players, know clearly why did we lost, won, or had a tie match!


```python
computer_choice = get_computer_choice()
player_choice = get_player_choice()
print(determine_winner(computer_choice, player_choice))
```
